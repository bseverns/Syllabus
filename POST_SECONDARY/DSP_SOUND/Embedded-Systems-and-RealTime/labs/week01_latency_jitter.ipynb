# Week 01 — Latency + jitter (measure your loop)

We treat loop timing as a signal. We compute jitter statistics and visualize spikes.


```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
```

```python
# If you have a CSV from host/serial_loop_logger.py, load it here.
csv_path = Path("..")/"export"/"loop.csv"
if csv_path.exists():
    data = np.loadtxt(csv_path, delimiter=",", skiprows=1)
    loop_us = data[:,1]
    print("samples:", len(loop_us))
    print("mean us:", loop_us.mean(), "std us:", loop_us.std(), "max us:", loop_us.max())
    plt.figure()
    plt.plot(loop_us[:2000])
    plt.title("Loop timing (first 2000 samples)")
    plt.xlabel("sample")
    plt.ylabel("loop (us)")
    plt.show()
else:
    print("No loop.csv yet. Run: python host/serial_loop_logger.py ...")
```
