# Week 04 — ADC stability (software filtering)

We compare EMA smoothing, median filters, and deadband for knob feel.


```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
```

```python
x = np.linspace(0, 1, 2000)
raw = x + 0.02*np.random.randn(len(x))
raw = np.clip(raw, 0, 1)

def ema(sig, a=0.15):
    out = np.zeros_like(sig)
    out[0] = sig[0]
    for i in range(1, len(sig)):
        out[i] = a*sig[i] + (1-a)*out[i-1]
    return out

sm = ema(raw, 0.12)
plt.figure()
plt.plot(raw, label="raw", alpha=0.5)
plt.plot(sm, label="ema", alpha=0.9)
plt.legend()
plt.title("EMA smoothing demo")
plt.show()
```
