# Week 01 — Signals as objects (time domain first)

We begin with signals as arrays: numbers over time. We will **plot** them, and when possible, **listen**.

**Learning goals**
- Generate basic waveforms
- Plot short windows (milliseconds)
- Hear/see the link between shape and timbre


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
import soundfile as sf

# Optional: live playback (may fail on some systems)
try:
    import sounddevice as sd
except Exception:
    sd = None
```

```python
sr = 48000
dur = 1.0
t = np.arange(int(sr*dur)) / sr

def sine(freq=440.0, amp=0.8, phase=0.0):
    return amp * np.sin(2*np.pi*freq*t + phase)

def square(freq=440.0, amp=0.8):
    return amp * np.sign(np.sin(2*np.pi*freq*t))

x_sine = sine(440)
x_square = square(440)

# Plot 20 ms
N = int(sr*0.02)
plt.figure()
plt.plot(t[:N]*1000, x_sine[:N], label="sine")
plt.plot(t[:N]*1000, x_square[:N], label="square", alpha=0.7)
plt.xlabel("Time (ms)")
plt.ylabel("Amplitude")
plt.title("20 ms window")
plt.legend()
plt.show()
```

```python
# Listen (optional)
if sd is not None:
    sd.play(x_sine, sr); sd.wait()
    sd.play(x_square, sr); sd.wait()
else:
    print("No live playback available. Export and listen in Audacity if needed.")
```
