# Week 04 — Difference equations (IIR filters) + stability

Feedback makes filters powerful—and risky.

**Learning goals**
- Implement a first-order IIR
- Sweep parameters
- Observe stability and resonance


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
import soundfile as sf

# Optional: live playback (may fail on some systems)
try:
    import sounddevice as sd
except Exception:
    sd = None
```

```python
sr = 48000
dur = 2.0
t = np.arange(int(sr*dur))/sr
# A click + noise burst
x = np.zeros_like(t)
x[0] = 1.0
x += 0.05*np.random.randn(len(x))

def iir_one_pole(x, a):
    """y[n] = (1-a)*x[n] + a*y[n-1]  where 0<=a<1 stable"""
    y = np.zeros_like(x)
    for n in range(1, len(x)):
        y[n] = (1-a)*x[n] + a*y[n-1]
    return y

for a in [0.2, 0.8, 0.98]:
    y = iir_one_pole(x, a)
    plt.figure()
    plt.plot(y[:2000])
    plt.title(f"Impulse-ish response, a={a}")
    plt.show()
```

```python
# Stability exploration: what happens when a > 1?
for a in [1.01, 1.1]:
    y = iir_one_pole(x, a)
    plt.figure()
    plt.plot(y[:500])
    plt.title(f"Unstable example, a={a}")
    plt.ylim(-5, 5)
    plt.show()
```
