# Week 02 — Operations + measurement

We transform signals and measure them.

**Learning goals**
- Time shift, reverse, scale
- Compute RMS / energy
- Peek at autocorrelation as shape matching


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
import soundfile as sf

# Optional: live playback (may fail on some systems)
try:
    import sounddevice as sd
except Exception:
    sd = None
```

```python
sr = 48000
dur = 0.5
t = np.arange(int(sr*dur)) / sr
x = np.sin(2*np.pi*220*t) * signal.windows.hann(len(t))

# Operations
shift = 200
x_shift = np.concatenate([np.zeros(shift), x[:-shift]])
x_rev = x[::-1]
x_scale = 0.4 * x

# Plot
plt.figure()
plt.plot(x[:800], label="x")
plt.plot(x_shift[:800], label="shifted")
plt.plot(x_rev[:800], label="reversed", alpha=0.7)
plt.legend()
plt.title("Operations (first 800 samples)")
plt.show()
```

```python
# Measurement: RMS and energy
def rms(sig):
    return np.sqrt(np.mean(sig**2))

def energy(sig):
    return np.sum(sig**2)

print("RMS:", rms(x), rms(x_scale))
print("Energy:", energy(x), energy(x_scale))
```

```python
# Autocorrelation (normalized)
corr = signal.correlate(x, x, mode="full")
corr = corr / np.max(np.abs(corr))
lags = np.arange(-len(x)+1, len(x))

plt.figure()
plt.plot(lags, corr)
plt.title("Autocorrelation (normalized)")
plt.xlabel("Lag (samples)")
plt.show()
```
