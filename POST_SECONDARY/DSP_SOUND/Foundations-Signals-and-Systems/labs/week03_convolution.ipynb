# Week 03 — Convolution (LTI systems in the discrete world)

Convolution turns an impulse response into a working filter.

**Learning goals**
- Compute convolution
- Design tiny kernels
- Interpret results by plot + listening


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
import soundfile as sf

# Optional: live playback (may fail on some systems)
try:
    import sounddevice as sd
except Exception:
    sd = None
```

```python
# Toy example (small sequences)
x = np.array([0, 1, 2, 3, 2, 1, 0], dtype=float)
h = np.array([0.25, 0.5, 0.25], dtype=float)  # simple smoother
y = np.convolve(x, h, mode="full")

plt.figure()
plt.stem(x, linefmt="C0-", markerfmt="C0o", basefmt="k-", label="x", use_line_collection=True)
plt.stem(np.arange(len(h)), h, linefmt="C1-", markerfmt="C1o", basefmt="k-", label="h", use_line_collection=True)
plt.title("Toy signals x and h")
plt.legend()
plt.show()

plt.figure()
plt.stem(y, use_line_collection=True)
plt.title("y = x * h (full)")
plt.show()
```

```python
# Audio example: moving average and echo tap
sr = 48000
dur = 1.0
t = np.arange(int(sr*dur))/sr
x = np.sin(2*np.pi*440*t)

# Moving average FIR
h_ma = np.ones(33)/33
y_ma = np.convolve(x, h_ma, mode="same")

# Echo tap (delayed copy)
delay = int(0.15*sr)
h_echo = np.zeros(delay+1)
h_echo[0] = 1.0
h_echo[-1] = 0.5
y_echo = np.convolve(x, h_echo, mode="same")

plt.figure()
plt.plot(x[:1000], label="x")
plt.plot(y_ma[:1000], label="moving avg", alpha=0.8)
plt.plot(y_echo[:1000], label="echo", alpha=0.8)
plt.legend()
plt.title("First 1000 samples")
plt.show()

out_dir = Path("..") / "export"
out_dir.mkdir(exist_ok=True)
sf.write(out_dir/"week03_ma.wav", y_ma, sr)
sf.write(out_dir/"week03_echo.wav", y_echo, sr)
print("Exported week03_ma.wav and week03_echo.wav")
```
