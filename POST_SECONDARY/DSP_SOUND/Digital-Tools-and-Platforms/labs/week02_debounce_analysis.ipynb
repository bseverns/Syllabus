# Week 02 — Debounce analysis

We look at button bounce as data, not as an annoyance.


```python
import csv
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
```

```python
# Load week02 log (capture with serial_logger.py)
p = Path("..")/"export"/"week02.csv"
rows = []
with open(p, "r", newline="") as f:
    r = csv.DictReader(f)
    rows = list(r)

# Expect raw lines like: t_ms,rawState,debouncedState
t = []
raw = []
deb = []
for row in rows:
    s = row["raw_line"]
    if "," not in s: 
        continue
    parts = s.split(",")
    if len(parts) < 3: 
        continue
    t.append(float(parts[0])/1000.0)
    raw.append(int(parts[1]))
    deb.append(int(parts[2]))

t = np.array(t); raw = np.array(raw); deb = np.array(deb)
print("Samples:", len(t))
```

```python
plt.figure()
plt.step(t, raw, where="post", label="raw")
plt.step(t, deb, where="post", label="debounced")
plt.ylim(-0.2, 1.2)
plt.legend()
plt.title("Button bounce vs debounced state")
plt.xlabel("time (s)")
plt.show()
```
