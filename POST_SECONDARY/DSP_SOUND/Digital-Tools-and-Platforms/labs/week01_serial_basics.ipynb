# Week 01 — Serial basics: seeing the board think

We treat Serial as a window into the device.

Goal: log data → plot it → explain it.


```python
import csv
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
```

```python
# Load a CSV captured by scripts/serial_logger.py
p = Path("..")/"export"/"week01.csv"
print("Expecting:", p.resolve())
rows = []
with open(p, "r", newline="") as f:
    r = csv.DictReader(f)
    for row in r:
        rows.append(row)

print("Rows:", len(rows))
print("First 5:", rows[:5])
```

```python
# Try to parse numeric values from raw_line
vals = []
for row in rows:
    raw = row["raw_line"]
    try:
        vals.append(float(raw))
    except Exception:
        # maybe 't_ms,value'
        if "," in raw:
            parts = raw.split(",")
            try:
                vals.append(float(parts[1]))
            except Exception:
                pass

vals = np.array(vals, dtype=float)
print("Parsed numeric values:", vals.shape)
```

```python
plt.figure()
plt.plot(vals)
plt.title("Serial stream (parsed)")
plt.xlabel("sample index")
plt.ylabel("value")
plt.show()
```
